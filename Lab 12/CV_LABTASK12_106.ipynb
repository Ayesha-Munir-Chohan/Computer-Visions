```python
# Code Example 1: Auto Scaling in Medical Image Feature Analysis
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

data = pd.DataFrame({
    'Feature1': [10, 20, 30, 40, 50],
    'Feature2': [200, 190, 180, 170, 160]
})

scaler = StandardScaler()
scaled_data = scaler.fit_transform(data)

scaled_df = pd.DataFrame(scaled_data, columns=data.columns)

print("Original Data:\n", data)
print("\nScaled Data:\n", scaled_df)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.title("Original Features")
plt.plot(data['Feature1'], label='Feature1')
plt.plot(data['Feature2'], label='Feature2')
plt.legend()

plt.subplot(1, 2, 2)
plt.title("Scaled Features")
plt.plot(scaled_df['Feature1'], label='Feature1 (scaled)')
plt.plot(scaled_df['Feature2'], label='Feature2 (scaled)')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# Code Example 2: Detecting Cyclic Patterns in Industrial Sensor Data
import matplotlib.pyplot as plt
import pandas as pd

data = pd.Series([1, 2, 3, 4, 3, 2, 1, 2, 3, 4])

autocorrelation = [data.autocorr(lag=i) for i in range(1, len(data))]

plt.stem(range(1, len(data)), autocorrelation)

plt.title("Auto-Correlation Plot")
plt.xlabel("Lag")
plt.ylabel("Correlation")

plt.show()
```

```python
# Code Example 3: Identifying Repetitive Patterns in a Production Line Sensor
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)

base_pattern = [1, 2, 3, 4, 3, 2, 1, 2, 3, 4]
noise = np.random.normal(0, 0.2, len(base_pattern))

vibration_data = np.array(base_pattern) + noise

data = pd.Series(vibration_data)

autocorrelation = [data.autocorr(lag=i) for i in range(1, len(data))]

plt.figure(figsize=(10, 5))
plt.plot(data, marker='o', label='Vibration Data')

plt.title("Robotic Arm Vibration Signal (10 Seconds)")
plt.xlabel("Time (seconds)")
plt.ylabel("Vibration Level")

plt.grid(True)
plt.legend()

plt.show()

plt.figure(figsize=(10, 5))
plt.stem(
    range(1, len(data)),
    autocorrelation,
    basefmt=" "
)

plt.title("Autocorrelation of Vibration Signal")
plt.xlabel("Lag (seconds)")
plt.ylabel("Autocorrelation")

plt.grid(True)

plt.show()
```

```python
# Case Study 2: Seasonal Pattern Detection in Weather Data
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

days = 30

temperature = (
    20
    + 5 * np.sin(2 * np.pi * np.arange(days) / 7)
    + np.random.normal(0, 1, days)
)

temp_series = pd.Series(temperature)

autocorr_values = [
    temp_series.autocorr(lag=i)
    for i in range(1, 16)
]

plt.figure(figsize=(10, 5))

plt.stem(
    range(1, 16),
    autocorr_values,
    basefmt=" "
)

plt.title("Autocorrelation of Daily Temperature")
plt.xlabel("Lag (days)")
plt.ylabel("Autocorrelation")

plt.grid(True)

plt.show()
```

```python
# Case Study 3: Credit Scoring in Finance
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler

data = {
    'CreditHistoryScore': [650, 700, 620, 800, 720],
    'Income_k': [30, 45, 25, 70, 50],
    'Age': [25, 35, 22, 45, 33],
    'LoanApproved': [0, 1, 0, 1, 1]
}

df = pd.DataFrame(data)

X = df[['CreditHistoryScore', 'Income_k', 'Age']]
y = df['LoanApproved']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

model = LogisticRegression()
model.fit(X_scaled, y)

coefficients = pd.Series(
    model.coef_[0],
    index=X.columns
)

print("Feature Weights (Logistic Regression):")
print(coefficients.sort_values(ascending=False))
```

```python
# Case Study 1: Patient Health Risk Prediction (Healthcare)
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

data = {
    'BloodPressure': [120, 140, 130, 150, 110],
    'Cholesterol': [200, 240, 180, 260, 170],
    'HeartRate': [72, 88, 65, 95, 70],
    'Glucose': [90, 110, 85, 130, 80],
    'Risk': [0, 1, 0, 1, 0]
}

df = pd.DataFrame(data)

X = df[['BloodPressure', 'Cholesterol', 'HeartRate', 'Glucose']]
y = df['Risk']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

model = LogisticRegression()
model.fit(X_scaled, y)

print("Model Coefficients:")
print(model.coef_)
```

```python
# Case Study 2: Seasonal Crop Yield Forecasting (Agriculture)
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

months = 36

yield_data = (
    50
    + 10 * np.sin(2 * np.pi * np.arange(months) / 12)
    + np.random.normal(0, 2, months)
)

crop_series = pd.Series(yield_data)

autocorr_values = [
    crop_series.autocorr(lag=i)
    for i in range(1, 13)
]

plt.figure(figsize=(10, 5))

plt.stem(
    range(1, 13),
    autocorr_values,
    basefmt=" "
)

plt.title("Crop Yield Auto-Correlation")
plt.xlabel("Lag (Months)")
plt.ylabel("Auto-Correlation")

plt.grid(True)

plt.show()
```

```python
# Case Study 3: Creditworthiness Assessment (Finance)
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

data = {
    'Income': [30, 50, 25, 70, 45],
    'CreditHistory': [650, 720, 600, 800, 690],
    'EmploymentLength': [2, 5, 1, 10, 4],
    'LoanApproved': [0, 1, 0, 1, 1]
}

df = pd.DataFrame(data)

X = df[['Income', 'CreditHistory', 'EmploymentLength']]
y = df['LoanApproved']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

model = LogisticRegression()
model.fit(X_scaled, y)

weights = pd.Series(
    model.coef_[0],
    index=X.columns
)

print("Feature Weights:")
print(weights.sort_values(ascending=False))
```

```python
# Case Study 4: Anomaly Detection in Industrial Machines (Manufacturing)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler

np.random.seed(42)

vibration = [1, 2, 3, 4, 3, 2, 1, 2, 3, 7]
temperature = [40, 41, 42, 43, 42, 41, 40, 41, 42, 50]
torque = [10, 12, 14, 15, 14, 12, 10, 11, 13, 20]

df = pd.DataFrame({
    'Vibration': vibration,
    'Temperature': temperature,
    'Torque': torque
})

scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)

scaled_df = pd.DataFrame(
    scaled_data,
    columns=df.columns
)

autocorr_values = [
    scaled_df['Vibration'].autocorr(lag=i)
    for i in range(1, len(df))
]

plt.figure(figsize=(10, 5))

plt.stem(
    range(1, len(df)),
    autocorr_values,
    basefmt=" "
)

plt.title("Machine Vibration Auto-Correlation")
plt.xlabel("Lag")
plt.ylabel("Auto-Correlation")

plt.grid(True)

plt.show()
```

```python
# Case Study 5: Intrusion Detection in Network Security (Cybersecurity)
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier

data = {
    'PacketSize': [500, 1500, 300, 2000, 100],
    'Duration': [2, 10, 1, 15, 1],
    'LoginSuccessRate': [0.9, 0.2, 0.95, 0.1, 0.98],
    'UnusualPortAccess': [0, 1, 0, 1, 0],
    'Intrusion': [0, 1, 0, 1, 0]
}

df = pd.DataFrame(data)

X = df.drop('Intrusion', axis=1)
y = df['Intrusion']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(X_scaled, y)

importance = pd.Series(
    model.feature_importances_,
    index=X.columns
)

print("Feature Importance:")
print(importance.sort_values(ascending=False))
```
